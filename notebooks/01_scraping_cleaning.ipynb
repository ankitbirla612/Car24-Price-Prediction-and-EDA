```python
links = soup_H.find_all("a")

print("Total links:", len(links))

for link in links:
    text = link.get_text(" ", strip=True)

    if "km" in text.lower():
        print("\nTEXT:", text)
        print("HREF:", link.get("href"))
        print("CLASS:", link.get("class"))
        print("-" * 100)
```

```python
for link in soup_H.find_all("a"):
    text = link.get_text(" ", strip=True)

    if any(x in text.lower() for x in ["km", "petrol", "diesel"]):
        print("\nTEXT:", text[:500])
        print("HREF:", link.get("href"))
        print("CLASS:", link.get("class"))
        print("=" * 100)
```
